Studi Kasus: Penilaian Kelas
Skenario:
Kamu memiliki data nilai siswa dalam bentuk list. Kamu diminta untuk membuat program yang melakukan tugas berikut:

Filter: Mengambil nilai yang lulus saja (nilai >= 70).

Statistik: Menghitung rata-rata dari nilai yang lulus tersebut.

Klasifikasi: Memberikan status (Predikat) untuk setiap nilai yang lulus dengan aturan:

Nilai 90-100: "A"

Nilai 80-89: "B"

Nilai 70-79: "C"

```
print("="*30)
print("Penilaian Mahasiswa")
print("="*30)

#proses input
daftar_nilai = [70,80,65,95,85]
daftar_lulus=[]

for i in daftar_nilai:
  if i>=70:
    daftar_lulus.append(i)

    if i>=90:
      predikat="A"
    elif i>=80:
      predikat="B"
    elif i>=70:
      predikat="C"

  print(f"nilai {i} lulus dengan predikat {predikat}")

#menentukan rata rata


```

Studi Kasus: Laporan Penjualan Toko
Kamu punya data daftar transaksi (dalam bentuk list harga). Tugasmu adalah:

Buat program untuk memfilter hanya transaksi yang nilainya di atas Rp50.000.

Hitung berapa total pendapatan dari transaksi yang sudah difilter tersebut.

Berikan status apakah toko tersebut "Ramai" (jika total pendapatan > 200.000) atau "Biasa" (jika tidak).

```
daftar_transaksi= [20000,50000,60000,55000,75000]
daftar_50000 = []

for i in daftar_transaksi:
  if i >=50000:
    daftar_50000.append(i)

total_pendapatan = sum(daftar_50000)

if total_pendapatan>=200000:
  status="Ramai"
else:
  status="Biasa"

print(f"daftar transaksi >=50K: {daftar_50000}")
print(f"total pendapatan: Rp {total_pendapatan:,}")
print(f"status: {status}")
```
