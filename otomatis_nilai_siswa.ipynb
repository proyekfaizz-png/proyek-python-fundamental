```python
# 1. INPUT DATA (Semua data siswa dikumpulkan di sini)
daftar_siswa = [
    {"nama": "Faiz", "nilai": [80, 90, 85, 15, 90, 300, 75]},
    {"nama": "Dani", "nilai": [90, 70, 85, 250, 20, 86, 50]},
    {"nama" : "Ridwan", "nilai" : [80,70,50,40,80,70,60]}
]

# 2. PROSES LOGIC AUTOMASI
# Loop Luar: Mengabsen siswanya satu per satu
for siswa in daftar_siswa:

    # Setiap kali ganti siswa, wadah analisis di bawah ini di-RESET jadi kosongan/0
    nilai_normal = []
    nilai_outlier = []
    total = 0
    jumlah = 0

    # Loop Dalam: Membedah angka nilai dari siswa yang sedang diabsen
    for nilai in siswa["nilai"]:
        if 50 <= nilai <= 100:
            nilai_normal.append(nilai)
            total += nilai
            jumlah += 1
        else:
            nilai_outlier.append(nilai)



    # Hitung rata-rata untuk siswa tersebut
    rata_rata = total / jumlah if jumlah > 0 else 0

    # 3. OUTPUT (Dicetak langsung di dalam loop luar agar muncul untuk SETIAP siswa)
    print("="*45)
    print(f"Nama Siswa   : {siswa['nama']}")
    print(f"Nilai Normal : {nilai_normal}")
    print(f"Nilai Outlier: {nilai_outlier}")
    print(f"Total Nilai  : {total}")
    print(f"Rata-rata    : {rata_rata:.2f}")
```
